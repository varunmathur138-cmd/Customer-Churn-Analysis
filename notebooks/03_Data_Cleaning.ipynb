## Outlier Analysis using IQR

```python
def check_outlier(df, column):
    q1=df[column].quantile(0.25)
    q3=df[column].quantile(0.75)
    IQR = q3-q1
    lower_bound = q1-1.5*IQR
    upper_bound = q3+1.5*IQR
    outliers = df[(df[column]<lower_bound)|(df[column]>upper_bound)]
    print(f"column:{column}")
    print(f"lower bound : {lower_bound}")
    print(f"upper bound : {upper_bound}")
    print(f"Number of outliers : {len(outliers)}")
    return outliers
    
```

```python
check_outlier(customers, "age")
```

```python
check_outlier(customers, "tenure_months")
```

```python
check_outlier(transactions, "amount")
```

```python
check_outlier(customer_usage, "data_usage_gb")
```

```python
check_outlier(customer_usage, "call_minutes")
```

```python
check_outlier(customer_usage, "sms_count")
```

```python
check_outlier(support_tickets,"resolution_days")
```

```python
check_outlier(support_tickets, "satisfaction_score")
```
